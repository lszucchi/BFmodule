```python
agora = datetime.now()
name = agora.strftime("T_Change_Measurements_%Y_%m_%d__%H_%M_%S")

Mix_Chamber_T= np.full((300),np.nan)

Mix_Chamber_R= np.full((300),np.nan)

Fields= np.full((300),np.nan)

Magnet_Controller.ramp_rate=20e-3
Magnet_Controller.ramp_to(-20e-3)

for i in range(len(Fields)):
    Mix_Chamber_T[i]=Elsa.GetT('tmixing')
    Mix_Chamber_R[i]=Elsa.GetTR('tmixing')
    Fields[i]=Magnet_Controller.field()  
    sleep(1)
    clear_output(wait=True)
    plt.plot(Fields,Mix_Chamber_R,'bo-')
    plt.pause(0.05)
    np.savez(name,Fields=Fields,Mix_Chamber_R=Mix_Chamber_R,Mix_Chamber_T=Mix_Chamber_T)
```
